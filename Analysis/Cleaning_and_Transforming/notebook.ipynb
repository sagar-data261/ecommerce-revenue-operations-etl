```python
df[['customer_id', 'order_id']] = df[['customer_id', 'order_id']].astype(str)
```

```python
df['total_price'] = df['quantity'] * df['unit_price']
df['discounted_price'] = df['total_price'] - (df['discount'] * df['total_price'])
df['discounted_price'] = df['discounted_price'].round(2)
df['order_date'] = pd.to_datetime(df['order_date'])

df[['total_price', 'discounted_price', 'revenue', 'order_date']].head()
```
